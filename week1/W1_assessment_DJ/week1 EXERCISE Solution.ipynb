and responds with an explanation. This is a tool that you will be able to use yourself during the course!

````python
# set up environment
import os
from dotenv import load_dotenv

# Load environment variables in a file called .env

load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

if not api_key:
    print("Error: OPENAI_API_KEY not found in environment variables.")
else:
    print("API key found and looks good so far!")
````

````python
# imports
from openai import OpenAI
client = OpenAI()
````

````python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
````

````python
# here is the question; type over this to ask something new
default_question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
question = input(f"Please enter your question [{default_question.strip()}]: ")

if not question.strip():
    question = default_question.strip()

print(question)
````

````python
# prompts
system_prompt = "You are a helpful technical assistant.Your job is to answer technical questions in a simple and easy-to-understand way.- Explain concepts step by step.- Avoid unnecessary technical jargon.- Use simple examples when helpful.- Provide code examples when needed and explain them briefly.- Keep the answer clear, practical, and concise.- If the question is unclear, ask for clarification.- Always respond in markdown.- Keep the entire response under 300 words."
user_prompt = "Please give a detailed explanation to the following question: " + question
````

````python
#  messages

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
]
````

````python
# Get gpt-4o-mini to answer, with streaming
from IPython.display import display, Markdown, update_display

response = ""
display_handle = display(Markdown(""), display_id=True)

stream = client.chat.completions.create(
    model=MODEL_GPT,
    messages=messages,
    stream=True
)

for chunk in stream:
    # print(chunk.choices[0].delta.content)
    response += chunk.choices[0].delta.content or ''
    response = response.replace("```","").replace("markdown", "")
    # display(Markdown(response), display_id=display_handle.display_id)
    update_display(Markdown(response), display_id=display_handle.display_id)
# print (stream)
# stream
````
